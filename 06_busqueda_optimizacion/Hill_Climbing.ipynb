## Hill Climbing

Hill Climbing es un algoritmo de búsqueda local. En lugar de explorar un camino completo hacia una meta conocida, comienza con una solución y trata de mejorarla utilizando soluciones vecinas.

## 1. Idea intuitiva

Imagine una montaña cubierta por neblina.

No podemos observar toda la montaña. Solamente podemos comparar nuestra posición actual con posiciones cercanas.

El procedimiento es:

1. Comenzar con una solución.
2. Generar vecinos.
3. Evaluar los vecinos.
4. Elegir uno mejor.
5. Movernos.
6. Repetir mientras exista mejora.

Por eso recibe el nombre **Hill Climbing** o **escalada**.

## 2. Conceptos

### Solución actual
La alternativa en la que estamos en este momento.

### Vecino
Una solución cercana que puede obtenerse mediante un pequeño cambio.

### Función de evaluación
Función que asigna un valor a una solución.

Si queremos **maximizar**, un valor mayor representa una mejor solución.

Si queremos **minimizar**, un valor menor representa una mejor solución.

## 3. Ejemplo de la presentación

Queremos maximizar:

$$
f(x)=-(x-4)^2+16
$$

En este ejemplo:

- `x` = solución que estamos evaluando.
- `f(x)` = calidad de esa solución.
- vecinos de `x` = `x-1` y `x+1`.

Comenzaremos en:

$$
x=1
$$

## 4. Resolución a mano: x = 1

Valor actual:

$$
f(1)=-(1-4)^2+16=7
$$

Vecino izquierdo:

$$
f(0)=-(0-4)^2+16=0
$$

Vecino derecho:

$$
f(2)=-(2-4)^2+16=12
$$

Como estamos maximizando:

$$
12 > 7
$$

Nos movemos de `x=1` a `x=2`.

## 5. Continuamos subiendo

En `x=2`:

$$
f(2)=12
$$

El vecino `x=3` produce:

$$
f(3)=15
$$

Nos movemos:

```text
2 → 3
```

Después:

$$
f(4)=16
$$

Nos movemos:

```text
3 → 4
```

En `x=4`, los vecinos `3` y `5` producen `15`.

Ninguno supera `16`.

**Resultado:**

```text
x = 4
f(x) = 16
```

## 6. Implementar la función en Python

````python
def funcion(x):
    return -(x - 4) ** 2 + 16

for x in range(0, 9):
    print(f"x={x} | f(x)={funcion(x)}")
````

## 7. Implementación de Hill Climbing

````python
def hill_climbing(inicio):
    actual = inicio
    recorrido = [actual]

    while True:
        izquierda = actual - 1
        derecha = actual + 1

        candidatos = [
            actual,
            izquierda,
            derecha
        ]

        mejor = max(
            candidatos,
            key=funcion
        )

        print(
            f"x={actual} | "
            f"f(x)={funcion(actual)} | "
            f"mejor vecino={mejor}"
        )

        if funcion(mejor) <= funcion(actual):
            return actual, recorrido

        actual = mejor
        recorrido.append(actual)
````

## 8. Ejecutar el ejemplo

````python
resultado, recorrido = hill_climbing(1)

print("\nRecorrido:", recorrido)
print("Mejor solución:", resultado)
print("Valor:", funcion(resultado))
````

### Resultado esperado

```text
Recorrido: [1, 2, 3, 4]
Mejor solución: 4
Valor: 16
```

El código reproduce el procedimiento manual: comparar la posición actual con sus vecinos y moverse solamente si existe una mejora.

## 9. Visualización de la función y el recorrido

La gráfica permite observar por qué el algoritmo se mueve hacia `x=4`.

````python
import matplotlib.pyplot as plt

xs = list(range(-2, 11))
ys = [funcion(x) for x in xs]

plt.figure(figsize=(8, 5))
plt.plot(xs, ys, marker="o")

rx = recorrido
ry = [funcion(x) for x in recorrido]
plt.scatter(rx, ry, s=80)

for x, y in zip(rx, ry):
    plt.annotate(str(x), (x, y))

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Hill Climbing")
plt.grid(True)
plt.show()
````

## 10. Problema: máximo local

Un **máximo local** es una solución mejor que sus vecinos inmediatos, pero que no necesariamente es la mejor de todo el espacio de búsqueda.

Por eso:

```text
mejor vecino local ≠ mejor solución global
```

Hill Climbing puede detenerse en una cima pequeña aunque exista una cima mayor en otra región.

## 11. Problema: meseta

Una **meseta** aparece cuando varias soluciones vecinas tienen el mismo valor.

Si ninguna es estrictamente mejor, una versión básica de Hill Climbing puede detenerse aunque todavía existan regiones interesantes más adelante.

Otras variantes pueden permitir movimientos laterales, reinicios aleatorios u otras estrategias.

## 12. Comparación conceptual con A*

### A*
- Trabaja con caminos.
- Tiene un inicio y una meta.
- Utiliza costo acumulado y heurística.
- Mantiene alternativas pendientes.

### Hill Climbing
- Trabaja con una solución actual.
- Examina vecinos.
- Se mueve si encuentra una mejora.
- Puede quedar atrapado en óptimos locales.

No son el mismo algoritmo aunque ambos utilizan información para orientar una búsqueda.

## 13. Reinicios: idea para reducir el problema de máximos locales

Una estrategia sencilla consiste en ejecutar Hill Climbing desde varios puntos iniciales.

Cada ejecución puede terminar en una solución diferente.

Después comparamos los resultados y conservamos el mejor encontrado.

````python
inicios = [-3, 0, 1, 7, 10]

resultados = []

for inicio in inicios:
    mejor, recorrido_local = hill_climbing(inicio)
    resultados.append((funcion(mejor), mejor))

mejor_valor, mejor_x = max(resultados)

print("\nMejor x encontrado:", mejor_x)
print("Mejor valor encontrado:", mejor_valor)
````
