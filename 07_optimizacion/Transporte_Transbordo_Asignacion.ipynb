## 1. Transporte — heurística del costo mínimo


```python
costos = [[4,8,8],[2,5,7]]
oferta = [70,50]
demanda = [40,30,50]

def transporte_costo_minimo(costos, oferta, demanda):
    oferta, demanda = oferta.copy(), demanda.copy()
    solucion = [[0]*len(demanda) for _ in oferta]

    while sum(demanda) > 0:
        opciones = [(costos[i][j], i, j)
                    for i in range(len(oferta))
                    for j in range(len(demanda))
                    if oferta[i] > 0 and demanda[j] > 0]
        costo, i, j = min(opciones)
        cantidad = min(oferta[i], demanda[j])
        solucion[i][j] += cantidad
        oferta[i] -= cantidad
        demanda[j] -= cantidad
        print(f"O{i+1} → D{j+1}: {cantidad} unidades | costo={costo}")
    return solucion

def calcular_costo(solucion, costos):
    return sum(solucion[i][j]*costos[i][j]
               for i in range(len(solucion))
               for j in range(len(solucion[i])))

solucion = transporte_costo_minimo(costos, oferta, demanda)
print("Solución:", solucion)
print("Costo total:", calcular_costo(solucion, costos))
```

La heurística de costo mínimo construye una solución factible inicial; por sí sola no demuestra optimalidad.


## 2. Transbordo


```python
grafo = {
    "S": {"A":2, "B":4},
    "A": {"B":1, "T":5},
    "B": {"T":2},
    "T": {}
}

def costo_ruta(grafo, ruta):
    return sum(grafo[ruta[i]][ruta[i+1]] for i in range(len(ruta)-1))

rutas = [["S","A","T"], ["S","B","T"], ["S","A","B","T"]]
for ruta in rutas:
    print(" → ".join(ruta), "| costo/unidad =", costo_ruta(grafo, ruta))
```

## 3. Asignación


```python
from itertools import permutations

personas = ["Ana","Luis","Marta"]
tareas = ["A","B","C"]
costos = [[9,2,7],[6,4,3],[5,8,1]]

mejor_costo = float("inf")
mejor_asignacion = None

for asignacion in permutations(range(len(tareas))):
    total = sum(costos[p][asignacion[p]] for p in range(len(personas)))
    print(asignacion, "→", total)
    if total < mejor_costo:
        mejor_costo, mejor_asignacion = total, asignacion

print("\nMejor asignación:")
for i, persona in enumerate(personas):
    print(persona, "→", tareas[mejor_asignacion[i]])
print("Costo total:", mejor_costo)
```
