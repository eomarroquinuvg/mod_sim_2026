# Algoritmo Genético
## CC3074 — Modelización y Simulación

Maximizar `f(x)=x²`, `0 ≤ x ≤ 31`, con cromosomas de 5 bits.


```python
import random
import matplotlib.pyplot as plt
random.seed(42)

def crear_individuo():
    return [random.randint(0, 1) for _ in range(5)]

def decodificar(individuo):
    return int("".join(map(str, individuo)), 2)

def fitness(individuo):
    return decodificar(individuo) ** 2
```

## Selección, cruce y mutación


```python
def seleccionar(poblacion):
    candidatos = random.sample(poblacion, 3)
    return max(candidatos, key=fitness)

def cruzar(p1, p2):
    punto = random.randint(1, 4)
    return p1[:punto] + p2[punto:], p2[:punto] + p1[punto:]

def mutar(individuo, probabilidad=0.1):
    individuo = individuo.copy()
    for i in range(len(individuo)):
        if random.random() < probabilidad:
            individuo[i] = 1 - individuo[i]
    return individuo
```

```python
TAM = 20
GENERACIONES = 50
poblacion = [crear_individuo() for _ in range(TAM)]
historial = []

for generacion in range(GENERACIONES):
    nueva = []
    while len(nueva) < TAM:
        p1, p2 = seleccionar(poblacion), seleccionar(poblacion)
        h1, h2 = cruzar(p1, p2)
        nueva.append(mutar(h1))
        if len(nueva) < TAM:
            nueva.append(mutar(h2))
    poblacion = nueva
    historial.append(fitness(max(poblacion, key=fitness)))

mejor = max(poblacion, key=fitness)
print("Mejor individuo:", mejor)
print("x =", decodificar(mejor))
print("Fitness =", fitness(mejor))

plt.plot(range(1, GENERACIONES+1), historial)
plt.xlabel("Generación")
plt.ylabel("Mejor fitness")
plt.title("Evolución del Algoritmo Genético")
plt.grid(True)
plt.show()
```
