## Vecino más cercano

La regla es visitar siempre la ciudad no visitada más cercana.


```python
distancias = {
    "A": {"B": 10, "C": 15, "D": 20},
    "B": {"A": 10, "C": 35, "D": 25},
    "C": {"A": 15, "B": 35, "D": 30},
    "D": {"A": 20, "B": 25, "C": 30}
}

def vecino_mas_cercano(distancias, inicio):
    actual = inicio
    visitados = {inicio}
    ruta = [inicio]
    costo_total = 0

    while len(visitados) < len(distancias):
        opciones = {c: d for c, d in distancias[actual].items() if c not in visitados}
        siguiente = min(opciones, key=opciones.get)
        print("Actual:", actual, "| Opciones:", opciones, "| Elegida:", siguiente)
        costo_total += opciones[siguiente]
        actual = siguiente
        visitados.add(actual)
        ruta.append(actual)

    costo_total += distancias[actual][inicio]
    ruta.append(inicio)
    return ruta, costo_total

ruta, costo = vecino_mas_cercano(distancias, "A")
print("\nRuta:", " → ".join(ruta))
print("Costo:", costo)
```

## Comparación con fuerza bruta


```python
from itertools import permutations

def costo_ruta(ruta, distancias):
    return sum(distancias[ruta[i]][ruta[i+1]] for i in range(len(ruta)-1))

inicio = "A"
otras = [c for c in distancias if c != inicio]
mejor_ruta, mejor_costo = None, float("inf")

for perm in permutations(otras):
    candidata = [inicio] + list(perm) + [inicio]
    costo = costo_ruta(candidata, distancias)
    if costo < mejor_costo:
        mejor_ruta, mejor_costo = candidata, costo

print("Óptimo:", " → ".join(mejor_ruta))
print("Costo óptimo:", mejor_costo)
```
