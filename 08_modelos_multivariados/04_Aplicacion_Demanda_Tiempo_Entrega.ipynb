# 04 - Aplicación: Demanda y Tiempo de Entrega

Variables:

- `X1` = demanda diaria
- `X2` = tiempo de entrega

Datos:

- media demanda = 100
- media entrega = 5
- desviación demanda = 20
- desviación entrega = 1
- correlación = 0.6

## Construcción de la matriz

`Cov(X1,X2) = r × s1 × s2`

`Cov(X1,X2) = 0.6 × 20 × 1 = 12`

Varianzas:

`Var(X1) = 20² = 400`

`Var(X2) = 1² = 1`

Por tanto:

`Σ = [[400, 12], [12, 1]]`

`μ = [100, 5]`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

media = [100,5]

covarianza = [
    [400,12],
    [12,1]
]

simulacion = np.random.multivariate_normal(
    mean=media,
    cov=covarianza,
    size=1000
)

datos = pd.DataFrame(
    simulacion,
    columns=["Demanda","Tiempo_Entrega"]
)

datos.head()
```

```python
print("Medias:")
print(datos.mean())

print("\nDesviaciones:")
print(datos.std())

print("\nCorrelación:")
print(datos.corr())
```

```python
plt.figure(figsize=(7,5))
plt.scatter(datos["Demanda"], datos["Tiempo_Entrega"], alpha=0.5)
plt.xlabel("Demanda diaria")
plt.ylabel("Tiempo de entrega")
plt.title("Demanda vs. Tiempo de entrega")
plt.grid(True)
plt.show()
```

## Interpretación

Como `r = 0.6`, esperamos que escenarios con mayor demanda tiendan también a presentar mayores tiempos de entrega.

La relación se observa en el conjunto de muchos escenarios, no necesariamente en cada observación individual.

## Escenario de riesgo

Definimos:

- demanda mayor a 120 unidades
- tiempo de entrega mayor a 6 días

```python
riesgo = datos[
    (datos["Demanda"] > 120) &
    (datos["Tiempo_Entrega"] > 6)
]

probabilidad = len(riesgo) / len(datos)

print("Escenarios de riesgo:", len(riesgo))
print("Total:", len(datos))
print("Probabilidad simulada:", probabilidad)
```

## Comparación con variables independientes

```python
np.random.seed(42)

demanda_ind = np.random.normal(100,20,1000)
tiempo_ind = np.random.normal(5,1,1000)

print(
    "Correlación al generarlas por separado:",
    np.corrcoef(demanda_ind, tiempo_ind)[0,1]
)
```
