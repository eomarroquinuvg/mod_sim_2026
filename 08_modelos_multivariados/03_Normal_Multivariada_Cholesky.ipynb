# 03 - Normal Multivariada y Cholesky

Una normal univariada:

`X ~ N(μ, σ²)`

Una normal multivariada:

`X ~ N(μ, Σ)`

Donde:
- `μ` = vector de medias
- `Σ` = matriz de covarianza

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

media = [100, 50]

covarianza = [
    [25, 10],
    [10, 16]
]

datos = np.random.multivariate_normal(
    mean=media,
    cov=covarianza,
    size=1000
)

print(datos[:5])
```

Cada fila representa una observación `[X1, X2]`.

```python
print("Media simulada X1:", datos[:,0].mean())
print("Media simulada X2:", datos[:,1].mean())

print("\nMatriz de covarianza simulada:")
print(np.cov(datos.T))
```

```python
plt.figure(figsize=(7,5))
plt.scatter(datos[:,0], datos[:,1], alpha=0.5)
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Distribución Normal Multivariada")
plt.grid(True)
plt.show()
```

## Generación sencilla de dos variables correlacionadas

Partimos de:

`Z1 ~ N(0,1)`

`Z2 ~ N(0,1)`

independientes.

Después:

`X1 = Z1`

`X2 = ρZ1 + sqrt(1 - ρ²)Z2`

```python
np.random.seed(42)

rho = 0.8
n = 1000

Z1 = np.random.normal(0,1,n)
Z2 = np.random.normal(0,1,n)

X1 = Z1
X2 = rho*Z1 + np.sqrt(1-rho**2)*Z2

print("Correlación simulada:", np.corrcoef(X1,X2)[0,1])
```

```python
plt.figure(figsize=(7,5))
plt.scatter(X1, X2, alpha=0.5)
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Correlación aproximada ρ = 0.8")
plt.grid(True)
plt.show()
```

## Descomposición de Cholesky

Buscamos `L` tal que:

`Σ = L × Lᵀ`

Después:

`X = μ + LZ`

`Z` contiene variables normales independientes y `L` introduce la estructura de covarianza.

```python
Sigma = np.array([
    [25,10],
    [10,16]
], dtype=float)

L = np.linalg.cholesky(Sigma)

print("L:")
print(L)

print("\nL @ L.T:")
print(L @ L.T)
```

```python
np.random.seed(42)

Z = np.random.normal(0,1,size=(2,1000))
mu = np.array([[100],[50]])

X = mu + L @ Z

print("Primeras 5 observaciones:")
print(X[:,:5].T)

print("\nCovarianza aproximada:")
print(np.cov(X))
```

## Idea clave

Cholesky transforma variables independientes en variables correlacionadas según una matriz de covarianza determinada.
