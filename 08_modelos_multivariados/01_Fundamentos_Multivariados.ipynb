# 01 - Fundamentos de Modelos Multivariados

Un modelo multivariado trabaja con **dos o más variables al mismo tiempo**.

Ejemplos:
- temperatura y humedad,
- clientes y ventas,
- demanda y tiempo de entrega.

La idea es estudiar cada variable y también cómo se relacionan entre sí.

## Vector de variables

Si:
- `X1` = temperatura
- `X2` = humedad
- `X3` = velocidad del viento

Entonces:

`X = [X1, X2, X3]`

Una observación puede ser:

`X = [25, 70, 12]`

```python
import numpy as np
import pandas as pd

observacion = np.array([25, 70, 12])

print("Vector:", observacion)
print("Temperatura:", observacion[0])
print("Humedad:", observacion[1])
print("Viento:", observacion[2])
```

## Datos multivariados

Ejemplo de una tienda:

| Día | Clientes X | Ventas Y |
|---:|---:|---:|
| 1 | 10 | 100 |
| 2 | 12 | 120 |
| 3 | 15 | 150 |
| 4 | 18 | 180 |
| 5 | 20 | 200 |

```python
datos = pd.DataFrame({
    "Clientes": [10, 12, 15, 18, 20],
    "Ventas": [100, 120, 150, 180, 200]
})

datos
```

## Vector de medias

`μ = [μ1, μ2, ..., μn]`

Para estos datos:

`μ = [15, 150]`

```python
vector_medias = datos.mean().to_numpy()

print("Vector de medias:", vector_medias)
```

## Varianza

Fórmula muestral:

`s² = Σ(xi - x̄)² / (n - 1)`

Ejemplo:

`X = [10, 12, 14]`

`x̄ = 12`

`(10-12)² + (12-12)² + (14-12)² = 4 + 0 + 4 = 8`

`s² = 8 / (3-1) = 4`

```python
x = np.array([10, 12, 14])

print("Media:", x.mean())
print("Varianza muestral:", x.var(ddof=1))
```
