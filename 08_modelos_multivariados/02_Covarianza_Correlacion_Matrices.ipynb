# 02 - Covarianza, Correlación y Matrices

En un modelo multivariado interesa saber cómo cambian las variables **conjuntamente**.

## Covarianza

`Cov(X,Y) = Σ[(xi - x̄)(yi - ȳ)] / (n - 1)`

Interpretación:
- positiva: tienden a aumentar juntas,
- negativa: una tiende a aumentar cuando la otra disminuye,
- cercana a cero: poca relación lineal.

### Ejemplo manual

`X = [1, 2, 3]`

`Y = [2, 4, 6]`

`x̄ = 2`

`ȳ = 4`

Productos:

`(-1)(-2) = 2`

`(0)(0) = 0`

`(1)(2) = 2`

Entonces:

`Cov(X,Y) = (2 + 0 + 2) / (3 - 1) = 2`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

X = np.array([1, 2, 3])
Y = np.array([2, 4, 6])

print("Covarianza:", np.cov(X, Y, ddof=1)[0, 1])
```

## Correlación

`r = Cov(X,Y) / (sX × sY)`

`-1 ≤ r ≤ 1`

- `r ≈ 1`: relación positiva fuerte
- `r ≈ 0`: relación lineal débil
- `r ≈ -1`: relación negativa fuerte

```python
print("Correlación:", np.corrcoef(X, Y)[0, 1])
```

## Matriz de covarianza

Para dos variables:

`Σ = [[Var(X1), Cov(X1,X2)], [Cov(X2,X1), Var(X2)]]`

La diagonal contiene varianzas.

```python
datos = pd.DataFrame({
    "Clientes": [10, 12, 15, 18, 20],
    "Ventas": [100, 120, 150, 180, 200]
})

print(datos.cov())
```

## Matriz de correlación

La diagonal contiene `1` porque cada variable tiene correlación perfecta consigo misma.

```python
print(datos.corr())
```

## Gráfico de dispersión

```python
plt.figure(figsize=(7,5))
plt.scatter(datos["Clientes"], datos["Ventas"])
plt.xlabel("Clientes")
plt.ylabel("Ventas")
plt.title("Clientes vs. Ventas")
plt.grid(True)
plt.show()
```

## Para analizar

¿La relación es positiva o negativa? ¿Qué ocurriría si al aumentar clientes disminuyeran las ventas?
